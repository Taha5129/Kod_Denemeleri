```python
aracTürü = {
    "Otomobil": 0.5,
    "Motorsiklet": 0.4
}

sehirOranlar = {
    "Ankara": 0.6,
    "İstanbul": 0.7,
    "İzmir": 0.5
}

temel_prim = 5000

def sigortaPrimHesaplama(aracTürü_dict, sehir_dict):
    try:
        arac = input("ARAÇ TİPİ (Otomobil/Motorsiklet): ")
        sehir = input("Şehir (Ankara/İstanbul/İzmir): ")
        
        if arac not in aracTürü_dict or sehir not in sehir_dict:
            print("Hatalı giriş yaptınız!")
            return
        
        sonuc = temel_prim * (1 + sehir_dict[sehir] * aracTürü_dict[arac])
        print("Sigorta primi:", sonuc)
    except Exception as e:
        print("Bir hata oluştu:", e)

sigortaPrimHesaplama(aracTürü, sehirOranlar)
```

```python
def daskPrim(sigorta):
    if sigorta == "Sigorta Primi":
        while True:
            try:
                Brüt = int(input("Brüt m2 giriniz: "))
                Birim = int(input("Birim m2 sayıyı giriniz: "))
                sigortaBedeli = Brüt * Birim
                print("Sigorta Bedeli:", sigortaBedeli)
                break
            except Exception as e:
                print("Sayısal değer giriniz:", e)

    elif sigorta == "Dask":
        while True:
            try:
                Brüt = int(input("Brüt m2 giriniz: "))
                Birim = int(input("Birim m2 sayıyı giriniz: "))
                tarifRiskKatsayısı = int(input("Tarife Risk Katsayısı giriniz: "))
                indirimler = float(input("İndirim Yüzdesi giriniz: "))
                vergiler = float(input("Vergi oranı giriniz: "))

                dask = (Brüt * Birim) * tarifRiskKatsayısı * (1 - indirimler) * vergiler
                print("Dask:", dask)
                break
            except Exception as e:
                print("Sayısal değer giriniz:", e)
```
