```python
tarifeRiskKatsayısı = {
    "1.derece": 0.05,
    "2.derece": 0.06,
    "3.derece": 0.07,
    "4.derece": 0.08,
    "5.derece": 0.09
}

indirimler = {
    "1_yıl": 0.10,
    "2_yıl": 0.15,
    "3_yıl": 0.20
}

vergiler = {
    "giderVergisi": 0.05
}

def daskPrim():
    print("İşlem seçiniz:")
    print("1 - Sigorta Primi")
    print("2 - Dask")
    secim = input("Seçiminizi giriniz (1/2): ")

    if secim == "1":  # Sigorta Primi
        try:
            Brüt = int(input("Brüt m2 giriniz: "))
            Birim = int(input("Birim m2 sayısını giriniz: "))
            sigortaBedeli = Brüt * Birim
            print("Sigorta Bedeli:", sigortaBedeli)
        except ValueError:
            print("Sayısal değer giriniz!")

    elif secim == "2":  # Dask
        try:
            Brüt = int(input("Brüt m2 giriniz: "))
            Birim = int(input("Birim m2 sayısını giriniz: "))

            # Tarife katsayısını dict'ten seç
            print("Mevcut tarifeler:", list(tarifeRiskKatsayısı.keys()))
            tarife_anahtar = input("Tarife risk derecesi giriniz (örn: 1.derece): ")

            if tarife_anahtar not in tarifeRiskKatsayısı:
                print("Hatalı tarife derecesi girdiniz!")
                return
            katsayı = tarifeRiskKatsayısı[tarife_anahtar]

            # İndirimi dict'ten seç
            print("Mevcut indirimler:", list(indirimler.keys()))
            indirim_anahtar = input("İndirim anahtarı giriniz (örn: 1_yıl): ")

            if indirim_anahtar not in indirimler:
                print("Hatalı indirim anahtarı girdiniz!")
                return
            indirim_oranı = indirimler[indirim_anahtar]

            vergi_oranı = vergiler["giderVergisi"]

            dask = (Brüt * Birim) * katsayı * (1 - indirim_oranı) * (1 + vergi_oranı)
            print("Dask Primi:", dask)

        except ValueError:
            print("Sayısal değer giriniz!")

    else:
        print("Geçersiz seçim")
```
